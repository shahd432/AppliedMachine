```python
import pandas as pd 
df=pd.read_csv('processed_telco.csv.xls')
df.head(10)
```

```python
pd.set_option('display.max_columns', None)
df.head(10)
```

```python
df.drop('customerID', axis=1, inplace=True)
```

```python
df.isnull().sum()
df.duplicated().sum()
df.info()
```

```python
columns=['gender', 'InternetService', 'Contract', 'PaymentMethod', 'TenureGroup','SpendingLevel']
for col in columns:
    print(f"{col}: {df[col].unique()}")
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df["gender"] = le.fit_transform(df["gender"])
```

```python
from sklearn.preprocessing import OrdinalEncoder
contract_order = [["Month-to-month", "One year", "Two year"]]
oe = OrdinalEncoder(categories=contract_order)
df["Contract"] = oe.fit_transform(df[["Contract"]])
tenure_order = [["New", "Mid-term", "Long-term"]]
oe2 = OrdinalEncoder(categories=tenure_order)
df["TenureGroup"] = oe2.fit_transform(df[["TenureGroup"]])
spending_order = [["Low", "Medium", "High"]]
oe3 = OrdinalEncoder(categories=spending_order)
df["SpendingLevel"] = oe3.fit_transform(df[["SpendingLevel"]])
```

```python
df = pd.get_dummies(df, columns=["InternetService", "PaymentMethod"], drop_first=True)
```

```python
bool_cols = df.select_dtypes(include="bool").columns
df[bool_cols] = df[bool_cols].astype(int)
```
